```python
import math
import textatistic
import requests
import wordcloud
import pandas as pd
import numpy as np

from textblob import TextBlob
from textatistic import Textatistic
from bs4 import BeautifulSoup

```

```python
#create the 3 variables for analysis
website1 = 'https://www.nytimes.com/live/2022/04/07/us/ketanji-brown-jackson-vote-scotus'  # save website address
response1 = requests.get(website1) # get http response through the internet
soup1 = BeautifulSoup(response1.content, 'html.parser') # process http response
text1 = soup1.get_text(strip=True) # remove html tags

website2 = 'http://www.macdillthunderbolt.com/original_site/rightstory3_082004.htm'  
response2 = requests.get(website2) 
soup2 = BeautifulSoup(response2.content, 'html.parser') 
text2 = soup2.get_text(strip=True) 


website3 = 'https://www.rogerebert.com/reviews/chariot-movie-review-2022'  
response3 = requests.get(website3)
soup3 = BeautifulSoup(response3.content, 'html.parser') 
text3 = soup3.get_text(strip=True)
```

```python
readability1 = Textatistic(text1)
text_blob1 = TextBlob(text1)

readability2 = Textatistic(text2)
text_blob2 = TextBlob(text2)

readability3 = Textatistic(text3)
text_blob3 = TextBlob(text3)
```

```python
#function performs polarity analysis of the 3 websites articles

def polarity_analysis1():
    polarityscore = text_blob1.sentiment.polarity
    polarity_score = float("{:.2f}".format(polarityscore))
    
    if polarity_score >0.35:
        print('With a Sentiment Polarity Score of', polarity_score, ', which is positive \U0001F600' )
    elif polarity_score > -.35 or polarity_score < .34:
        print('With a Sentiment Polarity Score of', polarity_score, ', which is neutral \U0001F610')
    elif polarity_score < -0.35:
        print('With a Sentiment Polarity Score of', polarity_score, ', which is negative \U0001F614')

def polarity_analysis2():      
    polarityscore2 = text_blob2.sentiment.polarity
    polarity_score2 = float("{:.2f}".format(polarityscore2))
    
    if polarity_score2 >0.35:
        print('With a Sentiment Polarity Score of', polarity_score2, ', which is positive \U0001F600' )
    elif polarity_score2 > -.35 or polarity_score2 < .34:
        print('With a Sentiment Polarity Score of', polarity_score2, ', which is neutral \U0001F610')
    elif polarity_score2 < -0.35:
        print('With a Sentiment Polarity Score of', polarity_score2, ', which is negative \U0001F614')

def polarity_analysis3():
    polarityscore3 = text_blob3.sentiment.polarity
    polarity_score3 = float("{:.2f}".format(polarityscore3))
                            
    if polarity_score3 > 0.35:
        print('With a Sentiment Polarity Score of', polarity_score3, ', which is positive \U0001F600' )
    elif polarity_score3 > -.35 or polarity_score3 < .34:
        print('With a Sentiment Polarity Score of', polarity_score3, ', which is neutral \U0001F610')
    elif polarity_score3 < -0.35:
        print('With a Sentiment Polarity Score of', polarity_score3, ', which is negative \U0001F614')
        
```

```python
def subjectivity_analysis1():
    subjectivityscore = text_blob1.sentiment.subjectivity
    subjectivity_score = float("{:.2f}".format(subjectivityscore))
    
    if subjectivity_score <0.5:
        print('The Sentiment Subjectivity Score is', subjectivity_score, ', which is mostly objective \n')
    elif subjectivity_score >.5:
        print('The Sentiment Subjectivity Score is', subjectivity_score, ', which is mostly subjective \n')
        
def subjectivity_analysis2():
    subjectivityscore2 = text_blob2.sentiment.subjectivity
    subjectivity_score2 =float("{:.2f}".format(subjectivityscore2))
    
    if subjectivity_score2 <0.5:
        print('The Sentiment Subjectivity Score is', subjectivity_score2, ', which is mostly objective \n')
    elif subjectivity_score2 >.5:
        print('The Sentiment Subjectivity Score is', subjectivity_score2, ', which is mostly subjective \n')

def subjectivity_analysis3():
    subjectivityscore3 = text_blob3.sentiment.subjectivity
    subjectivity_score3 = float("{:.2f}".format(subjectivityscore3))
    
    if subjectivity_score3 <0.5:
        print('The Sentiment Subjectivity Score is', subjectivity_score3, ', which is mostly objective ')
    elif subjectivity_score3 >.5:
        print('The Sentiment Subjectivity Score is', subjectivity_score3, ', which is mostly subjective')
        
```

```python
print('The text from:', website1)  
print('Has a Reability Dale-Chall score of', float("{:.2f}".format(readability1.dalechall_score)))
polarity_analysis1()
subjectivity_analysis1() 

print('The text from:',website2)  
print('Has a Reability Dale-Chall score of', float("{:.2f}".format(readability2.dalechall_score)))
polarity_analysis2()
subjectivity_analysis2() 

print('The text from:',website3)  
print('Has a Reability Dale-Chall score of', float("{:.2f}".format(readability3.dalechall_score)))
polarity_analysis3()
subjectivity_analysis3() 
```
