```python
''' This program retrieves twitter information from user-entered handle'''

import tweepy
import geopy
import keys # import the keys.py file to bring in bearer_token

#Authenticate with Twitter
client = tweepy.Client(bearer_token =keys.bearer_token)
print('Connected to Twitter successfully')

#create user input for twitter handle
user = input('Please enter a Twitter handle: ')
#call user handle and get meta data
acct = client.get_user(username=user, user_fields=['description', 'public_metrics'])

# print user data
print(f'Account id number:\t {acct.data.id}')
print(f'Name on Account:\t {acct.data.name}')
print(f'Account Twitter handle:\t {acct.data.username}')
print(f'\nDescription: {acct.data.description}\n')
print(f"Followers:\t\t {acct.data.public_metrics['followers_count']}")
print(f"Following:\t\t {acct.data.public_metrics['following_count']}")
print(f"Tweets:\t\t\t {acct.data.public_metrics['tweet_count']}\n==========================================================\n")

# retrieve tweet information of last 10 tweets
tweets = client.get_users_tweets(id=acct.data.id, tweet_fields=['author_id','context_annotations', 'created_at'], max_results=10)

for tweet in tweets.data:
    if tweet.context_annotations == 0:
         print('Tweet Author ID: ',tweet.author_id,'\t\tTime of Tweet: ',tweet.created_at,'\nTweet: ',tweet.text, '\n\nMeta Data:\n', tweet.context_annotations,'\n==============================[...]
    
    else:
        print('Tweet Author ID: ',tweet.author_id,'\t\tTime of Tweet: ',tweet.created_at,'\nTweet: ',tweet.text, '\n\nMeta Data:\n No Context Annotations Data Found\n==============================[...]

```
